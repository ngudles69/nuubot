### Objective

Nuubot.setup() centralizes setup on the Nuubot object.

It performs the following:

- loads config and validates via pydantic
- loads credentials and validates via pydantic
- loads datastore

### Usage

nuubot = Nuubot().setup()
print(f'mode: {nuubot.config.general.mode}')


```python
from nuubot import Nuubot
nuubot = Nuubot().setup()

try:
    print()
    print(f"Mode: {nuubot.config.general.mode}")
    print("Credentials: loaded")
    print("Datastore: initialized")
    print("Config:")
    print()
    print(nuubot.config.model_dump_json(indent=2))

    assert nuubot.config is not None
    assert nuubot.config.credentials is not None
    assert nuubot.datastore is not None
finally:
    nuubot.stop()
```
